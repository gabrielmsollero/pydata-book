## Generators

```python
some_dict = {"a": 1, "b": 2}

for key in some_dict:
    print(key)
```

When we iterate through the `dict` in a `for` loop, internally python calls the `iter` function with it - which would be the same as:

```python
dict_iterator = iter(some_dict)
dict_iterator
```

```python
list(dict_iterator)
```

We can create our own generator as follows:

```python
def squares(n = 10):
    print(f"Generating squares from 1 to {n}")
    for i in range(1, n + 1):
        yield i ** 2

# or, using generator expressions:
gen2 = (x ** 2 for x in range(10))


gen = squares()
gen
```

```python
for x in gen:
    print(x, end=" ")
```

Because we use `yield` instead of `return`, the next element won't be generated until the current one is consumed - thus saving some memory.

## The `itertools` module

Example: the `groupby` function:

```python
from itertools import groupby

def first_letter(s):
    return s[0]

names = ["Bibi", "Juju", "Jose", "Manu", "Bruno"]

for letter, names in groupby(names, first_letter):
    print(letter, list(names))
```

Some other useful `itertools` functions are:

| Function | Description |
| -------- | ----------- |
| `chain(*iterables)` | Chains iterators together, generating a single sequence |
| `combinations(iterable, k)` | Generates a sequence of all possible tuples with `k` elements using the iterator |
| `permutation(iterable, k)` | Same as above, but respecting order |
| `cartesian(*iterables, repeat=1)` | Cartesian (?) product of input iterables |
