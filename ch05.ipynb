## Introduction to Pandas

### Series


```python
import numpy as np
import pandas as pd

obj = pd.Series([4, 7, -5, 3])  # Series are 1D
obj
```

```python
# indexing each point with a label:
obj2 = pd.Series([4, 7, 5, -3], index=["d", "b", "a", "c"])
obj2
```

```python
obj2["a"]  # 5
obj2[["c", "a", "d"]]
```

```python
# numpy-like operations preserve the index-value link:
obj2[obj2 > 0]
obj2 * 2
```

```python
# can also be treated as a dictionary:
"b" in obj2
```

```python
# in fact a Series can be created from a dict:
sdata = {"Berlandia": 700000, "Araguari": 50000}
obj3 = pd.Series(sdata)
obj3
```

```python
# and vice-versa:
obj3.to_dict()
```

```python
# we can also pick indexes from the dict - order from index param will be respected and missing values will be replaced with NaN:
cities = ["Araguari", "Beraba", "Berlandia"]
obj4 = pd.Series(sdata, index=cities)
obj4
```

```python
# isna and notna can be used to detect missing data:
obj4.isna()  # same as pd.isna(obj4)
obj4.notna()
```

```python
# operations between series use labels to determine matching elements:
obj3 + obj4
```

### Dataframes

Can think of it as a dictionary of Series all sharing the same index


```python
# most common way to initialize is from a dict of equal length lists:
data = {
    "state": ["Ohio", "Ohio", "Ohio", "Nevada", "Nevada", "Nevada"],
    "year": [2000, 2001, 2002, 2001, 2002, 2003],
    "pop": [1.5, 1.7, 3.6, 2.4, 2.9, 3.2],
}
frame = pd.DataFrame(data)
frame
```

```python
# get only first few or last few rows - useful for large dfs:
frame.head()
frame.tail()
```

```python
# column rearranging and adding blank columns also works just like in Series:
frame2 = pd.DataFrame(data, columns=["year", "state", "pop", "non_existing"])
frame2
```

```python
# columns can be retrieved as series:
frame2["state"]
frame2.year
```

```python
# retrieving rows by position or name:
frame2.loc[1]
frame2.iloc[1]
```

```python
# assign a scalar to a column - all rows will be changed:
frame2["debt"] = 16.5
frame2
```

```python
# assigning an array:
frame2["debt"] = np.arange(6)
# frame2["debt"] = np.arange(7) # this would raise ValueError because lengths don't match
frame2
```

```python
# assigning a Series - labels will be matched and missing will be replaced by NaN
val = pd.Series([1, 2, 3], index=[2, 4, 5])
frame2["debt"] = val
frame2
```

```python
# create and delete columns:
frame2["eastern"] = frame2["state"] == "Ohio"
frame2
```

```python
del frame2["eastern"]
frame2
```

```python
# another common input data: nested dictionaries - inner keys will be treated as row indices
populations = {"Ohio": {2000: 1.5, 2001: 1.7, 2002: 3.6}, "Nevada": {2001: 2.4, 2002: 2.9}}
frame3 = pd.DataFrame(populations)
frame3
```

```python
# can transpose just like in numpy:
frame3.T
```

```python
# same works for dictionaries of series:
pdata = {"Ohio": frame3["Ohio"][:-1], "Nevada": frame3["Nevada"][:2]}
pd.DataFrame(pdata)
```

```python
# we can give names to index and columns:
frame3.index.name = "year"
frame3.columns.name = "state"

frame3  # frame3.T will transpose names too
```

```python
# can convert to ndarray:
frame2.to_numpy()
```

## Essential functionality

### Reindexing


```python
obj = pd.Series([4, 7, -5, 3], index=["d", "b", "a", "c"])
obj
```

```python
obj2 = obj.reindex(["a", "b", "c", "d", "e"])  # missing values are filled with NaN
obj2
```

```python
# use case: interpolation/filling:
obj3 = pd.Series(["blue", "purple", "yellow"], index=[0, 2, 5])
obj3
```

```python
obj3.reindex(
    np.arange(6), method="ffill"
)  # create a new series with indexes 1 to 5; will always use last value read from obj3 when reindexing
```

```python
# can reindex columns too:
frame = pd.DataFrame(np.arange(9).reshape((3, 3)), index=["a", "c", "d"], columns=["MG", "SP", "RJ"])
frame
```

```python
frame.reindex(columns=["PB", "MG", "RJ"])
```

```python
# can also reindex using loc:
frame.loc[["a", "d", "c"], ["RJ", "MG"]]  # will raise with PB because it's a null column
```

### Indexing, selection & filtering

Essentially the same as NumPy's, but you can also use index values (not just integers)


```python
obj1 = pd.Series([1, 2, 3])

obj1[[0, 1]]
obj1.loc[[0, 1]]  # same thing
```

```python
# when indexes are not integers, the first will work (with a warning) but the second won't:
obj2 = pd.Series([1, 2, 3], index=["a", "b", "c"])
obj2[[0, 1]]
```

```python
obj2.loc[[0, 1]]
```

This is why it's recommended to always index with `loc`.


```python
# iloc will work exclusively with integers:
obj2.iloc[[0, 1]]  # get first and second items, regardless of their labels
# obj2.iloc[['a', 'b']] # throws ValueError
```

```python
# slicing with labels is also possible
obj2.loc["b":"c"]
```

**IMPORTANT:** For label slicing, the **endpoint is inclusive**


```python
# assignment works with loc/iloc too:
obj2.loc["b":"c"] = 5
obj2
```

```python
data = pd.DataFrame(
    np.arange(16).reshape((4, 4)),
    index=["Ohio", "Colorado", "Utah", "New York"],
    columns=["one", "two", "three", "four"],
)
data
```

```python
data[["two", "three"]]  # selects columns
```

```python
data[:2]  # selects first two rows, all columns
```

```python
data[data["three"] > 5]  # select all rows where the column 'three' is greater than 5
```

```python
data[data < 5]  # replaces all the falsy with NaN
```

```python
data[data < 5] = 0  # assigns only to cells where the condition is true
data
```

#### loc and iloc in dataframes


```python
data.loc["Colorado", "two"]  # row, column
# data.loc["Colorado", ["two", "three"]] # multiple columns from a single row (as a series)
# data.loc[["Colorado", "Ohio"], ["two", "three"]] # multiple rows and columns (as a dataframe)
```

```python
# equivalents using iloc:
data.iloc[1, 1]
data.iloc[1, [1, 2]]
data.iloc[[1, 0], [1, 2]]

# and an extra:
data.iloc[:, :3][
    data["three"] > 5
]  # all rows, all columns except last; then filter rows where column 'three' is greater than 5
```

**IMPORTANT:** boolean arrays cannot be used with iloc


```python
# there's also at and iat that will return the value at a specific row and column
data.at["Colorado", "two"]
```

```python
# attention point when assigning with chained selections:
data.loc[data.three == 6]["three"] = 7  # won't work
```

```python
data.loc[data.three == 5, "three"] = 6  # works
data
```

### Arithmetic and data alignment

Both rows and columns are aligned when doing operations between two series or DFs. Non-matching indexes will be replaced in result by NaN or `fill_value`, if specified.


```python
# broadcasting in numpy:
arr = np.arange(12).reshape((3, 4))
arr - arr[0]  # element-wise subtraction of arr[0] is applied to every row in arr:
```

```python
# similarly, in pandas:
frame = pd.DataFrame(np.arange(12).reshape((4, 3)), columns=list("bde"), index=["Utah", "Ohio", "Texas", "Oregon"])
frame
```

```python
frame - frame.iloc[0]
```

```python
series = frame.iloc[0].copy()
series["a"] = 10
series
```

```python
frame - series  # column "a" will be created
```

```python
# can broadcast over columns too by specifying axis argument:
frame.sub(frame["b"], axis="index")
```

#### Mapping


```python
# numpy ufuncs work in dfs too:
frame = pd.DataFrame(np.arange(12).reshape((4, 3)), columns=list("abc"))
np.abs(frame)
```

```python
# we can apply any function to the df rows or columns using apply:
def f1(x):
    return x.max() - x.min()


frame.apply(f1)  # applies to columns
frame.apply(f1, axis="columns")  # applies to rows
```

```python
# functions applied can also return series:
def f2(x):
    return pd.Series([x.min(), x.max()], index=["min", "max"])


frame.apply(f2, axis="columns")
```

```python
# we can apply element-wise using map:
def my_format(x):
    return f"{x:.2f}"


frame.map(my_format)
```

#### Correlation and covariance


```python
price = pd.read_pickle("examples/yahoo_price.pkl")
volume = pd.read_pickle("examples/yahoo_volume.pkl")
```

```python
returns = price.pct_change()
returns.tail()
```

```python
# calculate correlation between percentage change in MSFT and IBM:
returns["MSFT"].corr(returns["IBM"])  # values are aligned by index
```

```python
# same for covariance
returns["MSFT"].cov(returns["IBM"])
```

```python
# then for the whole df, as a correlation matrix:
returns.corr()
```

```python
# calculate correlation for each column with a separate series:
returns.corrwith(returns["IBM"])
```

```python
# calculate correlation between matching columns of two dataframes:
volume["non_existing"] = 0
returns.corrwith(volume)
```

```python
# specifying axis="columns" calculates row by row
returns.corrwith(volume, axis="columns")
```
